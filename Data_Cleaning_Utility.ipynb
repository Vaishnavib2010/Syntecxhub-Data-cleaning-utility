```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/Diabetes Missing Data.csv")
```

```python
#Standardize Column Names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
#Handling Missing Values

#Detect Missing Values
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
#Drop Missing Values
df_drop = df.dropna()
```

```python
df_drop_col = df.dropna(axis=1)
```

```python
#Fill Missing Values (Imputation)
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].mean())
```

```python
#Fill with Median
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
#Fill with Mode (Categorical Columns)
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#Fill with Constant Value
df.fillna(0, inplace=True)
```

```python
df.dtypes
```

```python
#Fix Data Types
df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# Fix numeric columns
numeric_cols = [
    "Pregnant", "Glucose", "Diastolic_BP", "Skin_Fold",
    "Serum_Insulin", "BMI", "Diabetes_Pedigree", "Age", "Class"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Parse date columns automatically (if present)
for col in df.columns:
    if "date" in col.lower():
        df[col] = pd.to_datetime(df[col], errors="coerce")

# Verify datatypes
print(df.dtypes)
```

```python
#Remove duplicate rows
df = df.drop_duplicates()

#Standardize column names
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_", regex=False)
      .str.replace(r"[^a-z0-9_]", "", regex=True)
)

# Verify changes
print(df.shape)
print(df.columns)
```
